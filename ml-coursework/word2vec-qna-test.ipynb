```python
from gensim.models import Word2Vec

# Sample corpus
sentences = [["I", "love", "apples"], ["I", "love", "bananas"], ["Apples", "are", "tasty"], ["Bananas", "are", "tasty"]]

# Train Word2Vec model
model = Word2Vec(sentences, vector_size=3, window=5, min_count=1, workers=4)

# Get the embedding for a word
word_embedding = model.wv['apples']
print(f"Embedding for 'apples': {word_embedding}")
```

```python
from gensim.models import Word2Vec

# Sample corpus
sentences = [["I", "love", "apples"], ["I", "love", "bananas"], ["Apples", "are", "tasty"], ["Bananas", "are", "tasty"]]

# Train Word2Vec model
model = Word2Vec(sentences, vector_size=3, window=5, min_count=1, workers=4)

# Print embeddings for all unique words in the sentences
unique_words = set(word for sentence in sentences for word in sentence)
for word in unique_words:
    word_embedding = model.wv[word]
    print(f"Embedding for '{word}': {word_embedding}")
```
