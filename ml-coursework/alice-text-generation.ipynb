```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Bidirectional, Dropout, BatchNormalization, Dense
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import ReduceLROnPlateau, EarlyStopping
import numpy as np
from tqdm import tqdm
```

```python
print(f"Total unique words in the dataset: {total_words}")
print("Sample input sequences:")
for seq in input_sequences[:5]:
    print(seq)
```

```python
max_sequence_len = max([len(x) for x in input_sequences])
input_sequences = np.array(pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre'))

# Create predictors and labels
predictors, label = input_sequences[:, :-1], input_sequences[:, -1]

# Convert labels to one-hot encoded vectors
label = to_categorical(label, num_classes=total_words)
```

```python
for i in range(1):
    print(f"Label {i}: {label[i]}")
```

```python
print(f"Total unique words in the dataset: {total_words}")
print("Sample input sequences:")
for seq in input_sequences[:5]:
    print(seq)
```

```python
# Define and compile the model
model = Sequential()
model.add(Embedding(total_words, 200, input_length=max_sequence_len - 1))  # Increase embedding dimension
model.add(Bidirectional(LSTM(256, return_sequences=True)))  # Increase LSTM units
model.add(Dropout(0.3))  # Adjust dropout rate
model.add(BatchNormalization())
model.add(Bidirectional(LSTM(256)))
model.add(Dropout(0.3))
model.add(BatchNormalization())
model.add(Dense(total_words, activation='softmax'))
model.compile(loss='categorical_crossentropy', optimizer='RMSprop', metrics=['accuracy'])  # Change optimizer

# Print model summary
model.summary()
```

Certainly! Let’s break down the architecture and configuration of your neural network model:

Embedding Layer:
The Embedding layer converts integer-encoded word indices into dense vectors of fixed size (in this case, 200 dimensions).

It learns a representation for each word in the vocabulary.
The input shape for this layer is (None, 144) (where 144 is the maximum sequence length minus 1).

Bidirectional LSTM Layers:
Bidirectional LSTMs process input sequences in both forward and backward directions.

The first Bidirectional layer has 512 units and returns sequences (return_sequences=True).

The second Bidirectional layer has 512 units (no return_sequences).
These layers capture context from both past and future tokens.
Dropout Layers:
Dropout layers randomly set a fraction of input units to zero during training (here, with a dropout rate of 0.3).

This helps prevent overfitting by introducing some noise and promoting generalization.
Batch Normalization Layers:
Batch normalization normalizes the activations of the previous layer.
It helps stabilize training by reducing internal covariate shift.

Dense Layer (Output Layer):
The Dense layer produces the final output.
It has 552 units (equal to the total number of unique words in your dataset).
The activation function is softmax, which converts raw scores into probabilities.

Total Parameters:
The total number of parameters in the model is 2,908,520 (approximately 11.10 MB).

These parameters include weights and biases learned during training.
Trainable vs. Non-trainable Parameters:
Trainable parameters are updated during training to minimize the loss.
Non-trainable parameters (e.g., batch normalization statistics) remain fixed.
Remember that this architecture is designed for language modeling tasks.

Adjustments can be made based on your specific problem and dataset. Let me know if you need further clarification or have additional questions! 😊

```python
# Implement learning rate schedule and early stopping
reduce_lr = ReduceLROnPlateau(monitor='loss', factor=0.2, patience=5, min_lr=0.001)
early_stop = EarlyStopping(monitor='val_accuracy', patience=10, restore_best_weights=True)

# Use tf.data API for better performance
dataset = tf.data.Dataset.from_tensor_slices((predictors, label))
dataset = dataset.shuffle(buffer_size=len(predictors))
```

```python
# Split the dataset into training and validation sets
val_split = 0.1
val_size = int(len(predictors) * val_split)
val_dataset = dataset.take(val_size).batch(64, drop_remainder=True)
train_dataset = dataset.skip(val_size).batch(64, drop_remainder=True)

# Check the training dataset
print("Sample training dataset:")
for sample in train_dataset.take(5):
    print(sample)
```

```python
print(f"Validation dataset size: {val_size}")
print("Sample training dataset:")
for sample in train_dataset.take(5):
    print(sample)
print("Sample validation dataset:")
for sample in val_dataset.take(5):
    print(sample)
```

```python
# Train the model
history = model.fit(train_dataset, validation_data=val_dataset, epochs=100, callbacks=[reduce_lr, early_stop], verbose=1)

# Save the model weights
model.save('Alice-was_dataset_model.h5')
```
