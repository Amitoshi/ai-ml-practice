# Turn a Video into a Vertical Short

Cut a clip from a longer video and crop it to a 9:16 vertical format for Shorts/Reels using the ffmpeg command line. The original used a third-party AI shorts generator; this version keeps the core idea in a few simple steps.

## How to run
1. Run the install cell.
2. Upload a video you own.
3. Set the start time and length, then run the rest.

```python
from google.colab import files

uploaded = files.upload()
input_path = list(uploaded.keys())[0]
print("Uploaded:", input_path)
```

```python
START = "00:00:30"  #@param {type:"string"}
DURATION_SECONDS = 45  #@param {type:"integer"}
```

## Cut and crop
`crop=ih*9/16:ih` keeps the full height and a centered 9:16 slice; `scale=1080:1920` makes it standard Shorts size.

```python
import subprocess

output_path = "short.mp4"
cmd = [
    "ffmpeg", "-y",
    "-ss", START, "-i", input_path,
    "-t", str(DURATION_SECONDS),
    "-vf", "crop=trunc(ih*9/16/2)*2:ih,scale=1080:1920",
    "-c:v", "libx264", "-c:a", "aac",
    output_path,
]
subprocess.run(cmd, check=True)
print("Saved", output_path)
```
