# Compress Audio with ffmpeg

Shrink a large audio recording by converting it to mono at 64 kbps with ffmpeg. This is great for long voice recordings where stereo and high bitrate are not needed. Files are read from and saved to your Google Drive.

## How to run
1. Run the cells in order and allow Google Drive access.
2. Fill in the input file path and output folder in the form cell.

## Choose files
Paste the path of your audio file in Drive (right-click the file in the Colab file browser, then *Copy path*).

```python
INPUT_FILE = "/content/drive/MyDrive/input.mp3"  #@param {type:"string"}
OUTPUT_FOLDER = "/content/drive/MyDrive/compressed"  #@param {type:"string"}
BITRATE = "64k"  #@param ["32k", "64k", "96k", "128k"]
```

```python
import os
import subprocess

os.makedirs(OUTPUT_FOLDER, exist_ok=True)
name = os.path.splitext(os.path.basename(INPUT_FILE))[0]
output_file = os.path.join(OUTPUT_FOLDER, f"{name}_compressed.mp3")

# -ac 1 = mono, -b:a = audio bitrate, -y = overwrite
cmd = ["ffmpeg", "-y", "-i", INPUT_FILE, "-ac", "1", "-b:a", BITRATE, output_file]
subprocess.run(cmd, check=True)

before = os.path.getsize(INPUT_FILE) / 1e6
after = os.path.getsize(output_file) / 1e6
print(f"{before:.1f} MB -> {after:.1f} MB  ({output_file})")
```
