# MP3 Trimmer

Upload an MP3 and cut out a section of it (for example, remove the first 10 seconds) using `pydub`, which uses ffmpeg under the hood. The original tried an AI command-line tool; plain pydub is simpler and needs no API key.

## How to run
1. Run the install cell.
2. Run the upload cell and pick an MP3.
3. Set the start/end seconds and run the remaining cells.

## Upload your song

```python
from google.colab import files

uploaded = files.upload()
input_path = list(uploaded.keys())[0]
print("Uploaded:", input_path)
```

## Choose what to keep
Times are in seconds. Leave `END_SECONDS = None` to keep until the end.

```python
START_SECONDS = 10    # skip the first 10 seconds
END_SECONDS = None    # e.g. 60 to stop at 1 minute
```

```python
from pydub import AudioSegment

song = AudioSegment.from_file(input_path)
print(f"Original length: {len(song) / 1000:.1f} s")

start_ms = int(START_SECONDS * 1000)
end_ms = len(song) if END_SECONDS is None else int(END_SECONDS * 1000)
trimmed = song[start_ms:end_ms]   # pydub slices in milliseconds

output_path = "trimmed.mp3"
trimmed.export(output_path, format="mp3", bitrate="192k")
print(f"Saved {output_path} ({len(trimmed) / 1000:.1f} s)")
```
