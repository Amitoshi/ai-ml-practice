# Word2Vec Word Embeddings: A First Look

This notebook trains a tiny Word2Vec model with gensim on a handful of sentences and inspects the resulting word vectors. It shows why text should be lower-cased before training, how to read the embedding for a word, and how to find similar words. With such a small corpus the vectors are not meaningful; the goal is to learn the API.

## 1. Prepare a tiny corpus

Word2Vec is case-sensitive, so `Apples` and `apples` would be two different words. We lower-case everything first.

```python
from gensim.models import Word2Vec

raw_sentences = ["I love apples", "I love bananas", "Apples are tasty", "Bananas are tasty"]
sentences = [s.lower().split() for s in raw_sentences]
print(sentences)
```

## 2. Train the model

`workers=1` plus a fixed `seed` makes the result reproducible.

```python
model = Word2Vec(sentences, vector_size=5, window=5, min_count=1, workers=1, seed=42, epochs=200)
print("Vocabulary:", model.wv.index_to_key)
```

## 3. Inspect embeddings

```python
for word in model.wv.index_to_key:
    print(f"{word:>8}: {model.wv[word].round(3)}")
```

## 4. Similar words

```python
print("Most similar to 'apples':", model.wv.most_similar("apples", topn=3))
print("Similarity apples vs bananas:", round(float(model.wv.similarity("apples", "bananas")), 3))
```
