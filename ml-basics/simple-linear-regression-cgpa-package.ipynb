# Simple Linear Regression: CGPA vs Salary Package

This notebook fits a simple linear regression model that predicts a student's placement package (in LPA) from their CGPA. The data is a small synthetic dataset generated in code, so the notebook runs anywhere. It shows the full flow: explore, split, train, evaluate (MAE, RMSE, R²) and predict.

**How to run:** Open in Google Colab (CPU is fine) or any Jupyter environment and run all cells.

_Rewritten from my original Colab notebook: `Simple linear Regrations.ipynb`._

## 1. Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## 2. Create the dataset

We generate 200 students whose package grows roughly linearly with CGPA, plus some random noise.

```python
rng = np.random.default_rng(42)
cgpa = rng.uniform(5.0, 10.0, size=200).round(2)
package = (0.6 * cgpa - 1.0 + rng.normal(0, 0.3, size=200)).round(2)
df = pd.DataFrame({"cgpa": cgpa, "package": package})
df.head()
```

```python
plt.scatter(df["cgpa"], df["package"], alpha=0.6)
plt.xlabel("CGPA"); plt.ylabel("Package (LPA)")
plt.title("CGPA vs Package")
plt.show()
```

## 3. Split into train and test sets

```python
X = df[["cgpa"]]   # 2-D input (a DataFrame with one column)
y = df["package"]  # 1-D target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
print(len(X_train), "train rows,", len(X_test), "test rows")
```

## 4. Train the model

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
print("Slope (coef_):", lr.coef_[0])
print("Intercept:", lr.intercept_)
```

## 5. Evaluate

```python
y_pred = lr.predict(X_test)
print(f"MAE : {mean_absolute_error(y_test, y_pred):.3f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test, y_pred)):.3f}")
print(f"R^2 : {r2_score(y_test, y_pred):.3f}")

plt.scatter(X_test["cgpa"], y_test, label="actual")
line_x = pd.DataFrame({"cgpa": np.linspace(5, 10, 50)})
plt.plot(line_x["cgpa"], lr.predict(line_x), color="red", label="fitted line")
plt.xlabel("CGPA"); plt.ylabel("Package (LPA)"); plt.legend(); plt.show()
```

## 6. Predict for a new student

We pass a DataFrame with the same column name used in training, which avoids the scikit-learn feature-name warning.

```python
new_student = pd.DataFrame({"cgpa": [8.5]})
print(f"Predicted package for CGPA 8.5: {lr.predict(new_student)[0]:.2f} LPA")
# Same result by hand: package = slope * cgpa + intercept
print(f"By hand: {lr.coef_[0] * 8.5 + lr.intercept_:.2f} LPA")
```
